```python
import pandas as pd
from baselines.prompt_resolver import resolve_prompts

# 1) load run output row (already generated)
out_df = pd.read_parquet("../results/unified_baseline/mmlu_pro/vanilla/gpt_4o_mini/responses.parquet")
row = out_df[out_df["query_id"].astype(str) == "70"].iloc[0]   # change id

# 2) load source MMLU row to get options
src_df = pd.read_parquet("../datasets/processed/mmlu_pro.parquet")  # adjust path if needed
src_row = src_df[src_df["id"].astype(str) == str(row["query_id"])].iloc[0]

raw_options = src_row.get("options", "")
if isinstance(raw_options, list):
    options_text = "\n".join(f"{chr(65+i)}. {opt}" for i, opt in enumerate(raw_options))
elif hasattr(raw_options, "tolist") and not isinstance(raw_options, (str, bytes)):
    v = raw_options.tolist()
    if isinstance(v, list):
        options_text = "\n".join(f"{chr(65+i)}. {opt}" for i, opt in enumerate(v))
    else:
        options_text = "" if pd.isna(v) else str(v)
else:
    options_text = "" if pd.isna(raw_options) else str(raw_options)

# 3) rebuild prompts exactly like runner does
resolved = resolve_prompts(
    modality="vanilla",
    dataset="mmlu_pro",
    model_name="gpt-4o-mini",
    system_values={
        "model_name": "gpt-4o-mini",
        "output_contract": "Return final answer only.",
    },
    user_values={
        "dataset": "mmlu_pro",
        "query": row["query"],
        "options": options_text,
    },
)

print("=== SYSTEM PROMPT ===\n")
print(resolved.system_prompt)
print("\n=== USER PROMPT (sent to model) ===\n")
print(resolved.user_prompt)
```
