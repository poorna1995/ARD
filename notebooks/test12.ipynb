```python
import pandas as pd
import json
import re
import unicodedata


def normalize_response_answer(value: str, *, strict: bool = False) -> str:
    """Normalize answer payloads like '{"answer": "..."}' to comparable text."""
    if value is None:
        return ""

    text = str(value).strip()

    # Parse JSON payload if present: {"answer": "..."}
    try:
        parsed = json.loads(text)
        if isinstance(parsed, dict) and "answer" in parsed:
            text = str(parsed["answer"])
    except Exception:
        pass

    text = text.strip().lower()
    text = re.sub(r"\.{3,}$", "", text).strip()  # remove trailing ellipsis
    text = re.sub(r"\s+", " ", text)

    if strict:
        text = unicodedata.normalize("NFKD", text)
        text = "".join(ch for ch in text if not unicodedata.combining(ch))
        text = re.sub(r"[^\w\s]", " ", text)
        text = re.sub(r"\s+", " ", text).strip()

    return text


# Apply to dataframe
df_raw["response_answer_norm"] = df_raw["response_answer"].apply(normalize_response_answer)
df_raw["response_answer_norm_strict"] = df_raw["response_answer"].apply(
    lambda x: normalize_response_answer(x, strict=True)
)

df_raw[["response_answer", "response_answer_norm", "response_answer_norm_strict"]].head(25)
df = pd.read_parquet("../results1/unified_baseline/gaia/orchestrator_pipeline.parquet")
```
